```python
from flask import Flask, request, redirect, render_template_string
from pyngrok import ngrok


ngrok.set_auth_token("38IuU34N9a94gmLP0IvIGl3pPvU_3KDTEiDnczTp7SPScLFXo")

app = Flask(__name__)

posts = []

HTML = """
<!DOCTYPE html>
<html>
<head>
  <title>Social Media Platform</title>
</head>
<body style="font-family:Arial">

<h2>Mini Social Media Platform</h2>

<form method="POST">
  <input name="user" placeholder="Username" required><br><br>
  <textarea name="content" placeholder="Write a post..." required></textarea><br><br>
  <button type="submit">Post</button>
</form>

<hr>

<h3>All Posts</h3>
{% for p in posts %}
  <div style="border:1px solid #ccc; padding:10px; margin:10px;">
    <b>{{p.user}}</b><br>
    {{p.content}}<br><br>

    👍 Likes: {{p.likes}}
    <form method="POST" action="/like/{{loop.index0}}">
      <button>Like</button>
    </form>

    <form method="POST" action="/comment/{{loop.index0}}">
      <input name="comment" placeholder="Add comment" required>
      <button>Comment</button>
    </form>

    <ul>
      {% for c in p.comments %}
        <li>{{c}}</li>
      {% endfor %}
    </ul>
  </div>
{% endfor %}

</body>
</html>
"""

@app.route("/", methods=["GET","POST"])
def home():
    if request.method == "POST":
        posts.append({
            "user": request.form["user"],
            "content": request.form["content"],
            "likes": 0,
            "comments": []
        })
        return redirect("/")
    return render_template_string(HTML, posts=posts)

@app.route("/like/<int:id>", methods=["POST"])
def like(id):
    posts[id]["likes"] += 1
    return redirect("/")

@app.route("/comment/<int:id>", methods=["POST"])
def comment(id):
    posts[id]["comments"].append(request.form["comment"])
    return redirect("/")

public_url = ngrok.connect(5000)
print("OPEN THIS LINK:", public_url)

app.run(port=5000)
```
